```python
import pandas as pd
```

```python
train_path = "final_data/train/dico_nli_train_track1_participant_labeled.csv"

train_df = pd.read_csv(train_path)

train_df.head()
```

```python
print("Train shape:", train_df.shape)

print("\nKolonlar:")
print(train_df.columns.tolist())

print("\nVeri türleri:")
print(train_df.dtypes)
```

```python
train_df.head(10)
```

```python
train_df.sample(5, random_state=42)
```

```python
label_col = "label"

print(train_df[label_col].value_counts())
```

```python
label_percentages = train_df[label_col].value_counts(normalize=True) * 100

print(label_percentages.round(2))
```

```python
import matplotlib.pyplot as plt

train_df[label_col].value_counts().plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("DiCo-NLI Train Label Distribution")
plt.xlabel("Labels")
plt.ylabel("Number of Examples")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
for i in range(5):
    row = train_df.iloc[i]

    print("=" * 80)
    print("Text 1:", row["text1"])
    print("Text 2:", row["text2"])
    print("Label:", row["label"])
```
